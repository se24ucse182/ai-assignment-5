```python
# Bayesian Network Example
# Problem: Rain Prediction

# Install first if needed:
# !pip install pgmpy

from pgmpy.models import DiscreteBayesianNetwork
from pgmpy.factors.discrete import TabularCPD
from pgmpy.inference import VariableElimination

# ---------------- CREATE MODEL ----------------
# Cloudy -> Rain
# Rain -> WetGrass

model = DiscreteBayesianNetwork([
    ('Cloudy', 'Rain'),
    ('Rain', 'WetGrass')
])

# ---------------- DEFINE CPDs ----------------

# Probability of Cloudy
cpd_cloudy = TabularCPD(
    variable='Cloudy',
    variable_card=2,
    values=[
        [0.5],   # Not Cloudy
        [0.5]    # Cloudy
    ]
)

# Probability of Rain given Cloudy
cpd_rain = TabularCPD(
    variable='Rain',
    variable_card=2,
    values=[
        [0.8, 0.2],   # No Rain
        [0.2, 0.8]    # Rain
    ],
    evidence=['Cloudy'],
    evidence_card=[2]
)

# Probability of WetGrass given Rain
cpd_wetgrass = TabularCPD(
    variable='WetGrass',
    variable_card=2,
    values=[
        [0.9, 0.1],   # Dry Grass
        [0.1, 0.9]    # Wet Grass
    ],
    evidence=['Rain'],
    evidence_card=[2]
)

# ---------------- ADD CPDs ----------------

model.add_cpds(
    cpd_cloudy,
    cpd_rain,
    cpd_wetgrass
)

# ---------------- CHECK MODEL ----------------

print("Model Valid:", model.check_model())

# ---------------- INFERENCE ----------------

inference = VariableElimination(model)

# Query:
# Probability of Rain given Wet Grass

result = inference.query(
    variables=['Rain'],
    evidence={
        'WetGrass': 1
    }
)

# ---------------- OUTPUT ----------------

print("\nProbability Table:\n")

print(result)

print("\nProbability of Rain:")
print(result.values[1])

print("\nProbability of No Rain:")
print(result.values[0])
```
